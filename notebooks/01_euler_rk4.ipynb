# Euler and Runge–Kutta: an exact-solution experiment

Run cells in order. We solve $y\prime=-y$, $y(0)=1$ and compare with $e^{-t}$. See `docs/mathematics.md` for the derivation. This notebook is a walkthrough; committed reference data are generated by the script.

```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt
root = Path.cwd()
if not (root / "python").is_dir():
    root = root.parent
assert (root / "python" / "ode_methods.py").is_file(), "Open from the repository or notebooks directory"
sys.path.insert(0, str(root / "python"))
from ode_methods import solve
```

## Compare trajectories
Euler follows the tangent slope for one step. RK4 averages four carefully chosen slope samples.

```python
t, euler = solve(lambda t,y: -y, 1, 0, 2, 20, "euler")
_, rk4 = solve(lambda t,y: -y, 1, 0, 2, 20, "rk4")
fig, ax = plt.subplots(facecolor="white")
ax.set_facecolor("white")
ax.plot(t, np.exp(-t), "k-", label="Exact")
ax.plot(t, euler, "o--", label="Euler")
ax.plot(t, rk4, "s:", label="RK4")
ax.set(xlabel="Time t", ylabel="y(t)")
ax.grid(False); ax.legend(); plt.show()
```
